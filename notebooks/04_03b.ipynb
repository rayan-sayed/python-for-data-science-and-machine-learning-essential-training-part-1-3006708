```python
import numpy as np
from numpy.random import randn
import pandas as pd
from pandas import Series, DataFrame

import matplotlib.pyplot as plt
from matplotlib import rcParams
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 5,4
```

### Defining axes, ticks, and grids

```python
x = range(1,10)
y =[1,2,3,4,0,4,3,2,1]
fig = plt.figure()
ax = fig.add_axes([.1,.1,1,1])
ax.plot(x,y)
```

```python
x = range(1,10)
y =[1,2,3,4,0,4,3,2,1]
fig = plt.figure()   #blank figure at the beginning
ax = fig.add_axes([.1,.1,1,1])  #adding axes to the f

ax.set_xlim([1,9])  #limits from 1 to 9
ax.set_ylim([0,5])  #limits from 0 to 5 (instead of 4)

ax.set_xticks([0,1,2,3,4,5,6,8,9,10])    # 7 is missing on x axis
ax.set_yticks([0,1,2,3,4,5]) 

ax.plot(x,y)
```

```python
fig = plt.figure()
ax = fig.add_axes([.1,.1,1,1])  #adding axes to the figure

ax.set_xlim([1,9])  #limits from 1 to 9
ax.set_ylim([0,5])  #limits from 0 to 5 (instead of 4)

ax.grid() #adding grids to the figure

ax.plot(x,y)  
```

### Generating multiple plots in one figure with subplots 

```python
fig = plt.figure()
fig, (ax1,ax2) = plt.subplots(1,2)
ax1.plot(x)
ax2.plot(x,y)
```
