```python
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import DataFrame
```

```python
data = {'names':['steve', 'john', 'richard', 'sarah', 'randy', 'micheal', 'julie'],
        'age':[20, 22, 20, 21, 24, 23, 22],
        'gender':['Male', 'Male', 'Male', 'Female', 'Male', 'Male', 'Female'],
        'rank':[2, 1, 4, 5, 3, 7, 6]}

df=DataFrame(data)
df
```

### Matplotlib's Bar Chart

```python
plt.bar(df['names'], df['age'])
plt.xlabel('Names')
plt.ylabel('Age')
plt.title('Comparing Ages')
plt.show()
```

```python
plot = sns.barplot(data=df, x='names', y = 'age')
plot.set_title("Comparing Ages")
plt.show()
```

```python
plt.bar(df['names'],df['rank'])
plt.xlabel('Names')
plt.ylabel('Ranks')
plt.title('Comapring Ranks')
plt.show()
```

```python
sns.barplot(data=df, x='names' , y='rank')
plot.set_title("Comparing Ranks")
plt.show()
```

### Line Plot Matplotlib

```python
plt.plot(df['names'],df['rank'])
plt.xlabel('Names')
plt.ylabel('Ranks')
plt.title('Comparing Ranks')
plt.show()
```

### Line Plot Seaborn

```python
plot = sns.lineplot(data=df, x='names', y = 'rank')
plt.show()
```

### Pie Chart Matplotlib

```python
plt.pie(df['age'],labels=df['names'])
plt.title("Age comparison")
plt.show()
```

```python
plt.pie(df['rank'],labels=df['names'])
plt.title("Ranks comparison")
plt.show()
```

### Pie Chart Seaborn

```python
colors = sns.color_palette('pastel')[0:5]
plt.pie(df['age'], labels = df['names'], colors = colors)
plt.show()
```
